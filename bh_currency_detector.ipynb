```python
from dotenv import load_dotenv
import os

load_dotenv()
api = os.getenv("ROBOFLOW_API_KEY")
```

```python
!pip install roboflow
from roboflow import Roboflow
rf = Roboflow(api_key=api)
project = rf.workspace("alfadhul-almubarak").project("bahrain_currency_object_detector")
version = project.version(2)
dataset = version.download("yolo26")
```

```python
from ultralytics import YOLO
model = YOLO("yolo26l.pt") 
```

```python
results = model("Bahrain_Currency_Object_Detector-2/train/images/-_-0-_jpg.rf.11f4f7a8ba8cb74d60f8954db9a0b3ad.jpg")
```

```python
import matplotlib.pyplot as plt
plt.imshow(results[0].plot()[..., ::-1]);
```

```python
model.train(data='Bahrain_Currency_Object_Detector-2/data.yaml', epochs=5)
```
